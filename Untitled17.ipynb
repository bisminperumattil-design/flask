```python
def generate_fibonacci(max_value):

    sequence = []
    a, b = 0, 1
    while a <= max_value:
        sequence.append(a)
        a, b = b, a + b
    return sequence


limit = 100
fib_seq = generate_fibonacci(limit)
print(f"Fibonacci sequence up to {limit}: {fib_seq}")
```

```python
def rotate_list(lst, positions, direction='right'):
    if not lst:
        return lst

    n = len(lst)
    positions = positions % n

    if direction == 'right':
        return lst[-positions:] + lst[:-positions]
    elif direction == 'left':
        return lst[positions:] + lst[:positions]
    else:
        raise ValueError("Direction must be 'left' or 'right'")

original = [1, 2, 3, 4, 5]
print("Original:   ", original)
print("Rotate R 2:", rotate_list(original, 2, 'right'))
print("Rotate L 2:", rotate_list(original, 2, 'left'))
```
